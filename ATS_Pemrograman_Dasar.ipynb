```python
#Jawaban Soal Nomor 2.

daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

# A. Menghitung NIlai Harian
energi = daya_kw * jam
print(f"{energi} kWh")

# B. Hitung capacity factor
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100
print(f"{capacity_factor:.2f}%")
```

```python
#Jawaban Soal 6

suhu = float(input("masukkan suhu: "))
tekanan = float(input("masukkan tekanan: "))

if suhu >= 550 or tekanan >= 14:
  print ("STATUS: TRIP")

elif suhu > 520 and tekanan > 12:
  print ("STATUS: WARNING")

else:
    print ("STATUS: NORMAL")
```

```python
#Jawaban Soal 8

volume = 100

while volume > 0:
  volume -= 20
  print(f"Volume saat ini: {volume} liter")
  if volume == 20:
    print("WARNING: Hampir kosong!")

print ("TANGKI KOSONG - Pengosongan berhenti.")
```

Jadi pertama-tama program menetapkan volume awal tangki sebesar 100 liter. Program kemudian melakukan perulangan selama volume masih di atas 0, di mana pada setiap iterasi volume akan dikurangi sebanyak 20 liter, lalu mencetak sisa volume saat ini. Jika volume tersisa mencapai tepat 20 liter, program akan mencetak peringatan bahwa tangki hampir kosong. Proses pengosongan ini akan terus berulang hingga volume mencapai 0 dan perulangan berhenti, lalu program mencetak pesan bahwa tangki sudah kosong

```python
#Jawaban Soal 10

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0

for i, nilai in enumerate(data, start=1):
    try:
        suhu = float(nilai)
    except ValueError:
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break
    jumlah_valid += 1

    if suhu > 70.0:
        print(f"Warning: Titik {i} suhu bearing tinggi!")

print(f"Jumlah data valid: {jumlah_valid}")
```

**Penjelasan Soal Nomor 10**

Jadi pada program soal nomor 10 fungsi utamanya adalah untuk memantau data suhu bearing dari beberapa titik pengukuran, cara kerjanya menggunakan proses looping (for), fungsi penanganan kesalahan (try-except), seleksi kondisi (if), serta kontrol perulangan (break). Jadi pertama-tama program akan meng-enumerate (membaca data suhu satu persatu) yang dimulai dari titik ke-1 yakni "62.5". Program kemudian mencoba mengubah bentuk data tersebut menjadi angka desimal (float). Jika data bernilai valid dan tidak melebihi batas, jumlah data valid akan bertambah, dan jika suhu melebihi 70.0, program akan mencetak peringatan bahwa suhu bearing tinggi. Namun, ketika program membaca data ke-4 yakni "N/A" yang tidak bisa diubah menjadi angka, terjadi kesalahan (ValueError), sehingga program langsung mencetak pesan bahwa data tidak valid dan menghentikan seluruh pemeriksaan menggunakan perintah break.   

Operator yang digunakan di program ini adalah:

2. Operator Relasional (>)

```python
# Kodingan Bonus : Pemantauan Penggunaan Listrik Harian

penggunaan_listrik = [12.5, 8.0, 15.2, 22.5, 9.1]

print("PROGRAM PEMANTAUAN LISTRIK HARIAN")

for hari, kWh in enumerate(penggunaan_listrik, start=1):

    if kWh > 20.0:
        status = "Boros / Beban Tinggi"
    else:
        status = "Normal"

    print(f"Hari ke-{hari}: {kWh} kWh - Status: {status}")

print("PROGRAM SELESAI")
```

Lalu selain penggunaan 20 (kurang dari atau sama dengan 20) maka dinyatakan menjadi normal.
