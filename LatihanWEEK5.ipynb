```python
# Soal 1 Belanja Buku
harga_buku = 18500
jumlah_buku = 7
uang = 150000
total = harga_buku * jumlah_buku
diskon = total * 10 // 100
bayar = total - diskon



print("Total belanja:", total)
print("Diskon:", diskon)
print("Harus dibayar:", bayar)
print("Kembalian:", uang - bayar)
print("Iuran per orang:", bayar // 4)
print("Sisa pembagian:", bayar % 4)
print("Uang cukup:", uang >= bayar)
```

```python
# Soal 2 PLTS Atap Rumah
total_panelsurya = 12
daya_panel = 450
rata_penyinaran = 5
efisiensi_panel = 0.8 / 1000
energi_harian = total_panelsurya * daya_panel * rata_penyinaran * efisiensi_panel
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * 1444.70
kebutuhan = 25

print("Energi per hari:", energi_harian, "kWh")
print("Energi per bulan:", energi_bulanan, "kWh")
print("Nilai listrik: Rp", nilai_listrik)
print("Kekurangan per hari:", kebutuhan - energi_harian, "kWh")
print("Kebutuhan tercukupi:", energi_harian >= kebutuhan)
```

```python
# Soal 3 Data generator
daya_generator = 200
waktu_operasi = 8
konsumsi_solar = 0.25
harga_solar = 13500
kapasitas_tangki = 1000

arus = 20
hambatan = 0.5


energi = daya_generator * waktu_operasi
solar_per_hari = energi * konsumsi_solar
biaya_solar = solar_per_hari * harga_solar
hari_penuh = kapasitas_tangki // solar_per_hari
sisa_solar = kapasitas_tangki % solar_per_hari
rugi_kabel = arus ** 2 * hambatan
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print("Energi (kWh)   :", energi)
print("Solar (liter)  :", solar_per_hari)
print("Biaya (Rp)     :", biaya_solar)
print("Tangki cukup   :", hari_penuh, "hari")
print("Sisa solar (L) :", sisa_solar)
print("Rugi kabel (W) :", rugi_kabel)
print("Rugi tinggi?   :", rugi_tinggi)
```
