```python
# Latihan 1
# Menentukan tahun kabisat menggunakan operator logika

tahun = int(input("Masukkan tahun: "))

kabisat = (tahun % 400 == 0) or (
    tahun % 4 == 0 and tahun % 100 != 0
)

print("HASIL PEMERIKSAAN TAHUN")
print(f"Tahun yang diperiksa : {tahun}")
print(f"Apakah tahun kabisat?: {kabisat}")
```

```python
# Latihan 2
# Membuat Program sendiri "Menghitung luas dan keliling persegi panjang"

panjang = float(input("Masukkan panjang persegi panjang: "))
lebar = float(input("Masukkan lebar persegi panjang: "))

luas = panjang * lebar
keliling = 2 * (panjang + lebar)

print("HASIL PERHITUNGAN")
print(f"Panjang  : {panjang}")
print(f"Lebar    : {lebar}")
print(f"Luas     : {luas:.2f}")
print(f"Keliling : {keliling:.2f}")
```
