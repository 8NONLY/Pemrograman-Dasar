```python
# Hands-on Praktikum 3
# Konversi total detik menjadi jam, menit, dan detik

total_detik = int(input("Masukkan total detik: "))

jam = total_detik // 3600
sisa_detik = total_detik % 3600

menit = sisa_detik // 60
detik = sisa_detik % 60

print("HASIL KONVERSI WAKTU")
print("Total detik :", total_detik)
print(f"Hasil konversi: {jam} jam, {menit} menit, {detik} detik")
```

```python
# Hands-on Praktikum 5
# Menghitung total harga setelah diskon

belanja = float(input("Masukkan total belanja: Rp"))

diskon = (0, 0.35 * belanja)[belanja > 1000000]
total_harga = belanja - diskon

print("RINCIAN PEMBAYARAN")
print(f"Total belanja  : Rp{belanja:,.2f}")
print(f"Jumlah diskon  : Rp{diskon:,.2f}")
print(f"Total harga    : Rp{total_harga:,.2f}")
```
